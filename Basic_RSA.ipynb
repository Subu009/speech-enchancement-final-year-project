# 🔐 Tiny Lock — RSA Challenge

RSA is an **asymmetric** encryption scheme. It uses two different keys:

| Key | Who has it | What it does |
|---|---|---|
| 🔓 **Public key** $(n, e)$ | Everyone | Encrypts (locks) |
| 🔑 **Private key** $d$ | Only the owner | Decrypts (unlocks) |

Anyone can lock a message with the public key, but only the private key can unlock it.

---

## Step 1 — Generate two secret primes

Pick two random primes $p$ and $q$, then multiply them:

$$n = p \times q$$

$n$ is **public**, but $p$ and $q$ stay **secret**.
⚠️ This challenge deliberately uses tiny **64-bit** primes, which is what makes the key breakable.

## Step 2 — Build the public key

The public exponent is almost always:

$$e = 65537$$

So the public key is the pair $(n, e)$.

## Step 3 — Encrypt the flag

Convert the flag to an integer $m$, then encrypt it:

$$c = m^e \bmod n$$

> ⚠️ $m$ must be **smaller than** $n$, otherwise decryption will not recover the flag.

---

## 🧾 The code

```python
from Crypto.Util.number import getPrime, bytes_to_long

p, q = getPrime(64), getPrime(64)   # Step 1: secret primes
n = p * q                           #         public modulus

e = 65537                           # Step 2: public exponent
public_key = (n, e)

flag = bytes_to_long(b"IBOH26{XXXXXXXXX}")   # Step 3: plaintext as integer
c = pow(flag, e, n)                         #         ciphertext
```

````python
from Crypto.Util.number import long_to_bytes

# To decrypt to get the flag, start by calculating the phi_n
phi_n = (p-1)*(q-1)

# Use phi_n to calculate the private_key
d = pow(e, -1, phi_n)

# Use the private_key to decrypt the ciphertext
m = pow(c, d, n)
m = long_to_bytes(m)

print(f'flag = {m}')

# Lastly, notice how both public key and private are different. This is what we call asymmetric encryption
print(f'public_key = {public_key}')
print(f'private_key = {d}')
````
